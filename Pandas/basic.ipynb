What is Pandas?

Pandas is a powerful Python library used for:

Data analysis
Data manipulation
Data cleaning
Working with tables and spreadsheets

It is mainly used in:

Data Science
Machine Learning
Data Analytics

Pandas provides fast and flexible data structures for handling structured data.



| Feature            | Description                              |
| ------------------ | ---------------------------------------- |
| Fast and Efficient | Handles large datasets quickly           |
| Easy Data Handling | Simple syntax for manipulating data      |
| File Support       | Works with CSV, Excel, JSON, SQL, etc.   |
| Data Cleaning      | Handles missing or duplicate data        |
| Powerful Analysis  | Supports grouping, filtering, statistics |






| Function     | Purpose                   |
| ------------ | ------------------------- |
| `head()`     | Shows first 5 rows        |
| `tail()`     | Shows last 5 rows         |
| `info()`     | Gives dataset information |
| `describe()` | Statistical summary       |
| `shape`      | Returns rows and columns  |
| `columns`    | Shows column names        |
| `dtypes`     | Shows data types          |
