# pie chart

```python
a = [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]
plt.pie(a, labels = label)
plt.show()
```

```python
a = [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]
plt.pie(a, labels = label , explode= [ 0,0.3,0,0])     # explode is used to alg krna
plt.show()
```

```python
a = [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]
plt.pie(a, labels = label , explode= [ 0,0.3,0,0]  , shadow = True)  # shadow is used mark pie chart
plt.show()
```

```python
a = [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]

colour = ["pink","yellow","brown","red"]
plt.pie(a, labels = label ,colors = colour , explode= [ 0,0.3,0,0]  , shadow = True)  # shadow is used mark pie chart
plt.show()
```

```python
a = [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]

colour = ["pink","yellow","brown","red"]
plt.pie(a, labels = label ,colors = colour , explode= [ 0,0.3,0,0]  , shadow = True , autopct= "%1.1f%%")  # autopct 
                                                                     #is used give percent pie chart
plt.show()
```

```python
a= [1,2,3,10]
label = ["apple","mango","kiwi","chiku"]

colour = ["pink","yellow","brown","red"]
plt.pie(a, labels = label ,colors = colour , explode= [ 0,0.3,0,0]  , shadow = True , autopct= "%1.1f%%"  )
plt.legend()          #33 legend is used to side m data 
plt.show()
```
