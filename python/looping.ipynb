## Looping statment

In Python, loops are used to execute a block of code repeatedly.
There are two primary types of loops: for loops and while loops.

1. #### for loop : 
The for Loop is used to iterate over a sequence (like a list, tuple, string, or range). It is best when you know in advance how many times you want to run the code.
##### syntax: 
       for x in variable_name:
        print(x)

2. #### while loop
Repeatedly executes a block of code as long as a specified condition remains True. It is useful when the number of iterations is not fixed.



```python
name = "muskan"
for x in name:
    print(x)
```

```python
a = [1,2,3,4]
for x in a:
    print(x)
```

```python
def even(lst):
    print(lst)
    for x in lst:
        if x%2==0:
             print("x is even number:",x)
        else:
            print("x is odd number:",x)    
lst1 = [1,2,3,4,5,5]
even(lst=lst1)                   
```

```python
def grade(a):
    print(a)
    for x in a:
        if x<=100 and x>=90:
            print("A++")
        elif x <=90 and x>=80:
            print("A")
        elif x<=79 and x>=60:
            print("B")
        elif x<=59 and x>=40:
            print("C")
        else:
            print("d")
a= [22,45,78,98,65]
grade(a )                       

```
