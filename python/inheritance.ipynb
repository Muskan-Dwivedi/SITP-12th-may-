Inheritance in Python

Inheritance is an Object-Oriented Programming (OOP) concept where one class (child/derived class) acquires the properties and methods of another class (parent/base class).

It helps in:

Code reusability
Better organization
Easy maintenance
Extending existing functionality

# Types of Inheritance in Python


| **Type of Inheritance**  |	   **Description**	   |    **Example Structure**|
|--------------------------|---------------------------|------------------------|
| Single Inheritance	   |One child class inherits from one parent class  | A Dog inherits from Animal|
|Multiple Inheritance	|One child class inherits from multiple parent classes	|A, B → C
|Multilevel Inheritance	|A child class becomes parent for another class	|A → B → C
|Hierarchical Inheritance	|Multiple child classes inherit from one parent class	|A → B, C
|Hybrid Inheritance         |Combination of two or more inheritance types  |Mixed structure

```python
class Person:
    def person_info(a):
        print("i am person")
class Teacher(Person):
    def teacher_info(self):
        print("i am teacher")
class HOD(Teacher):
    def hod_info(b):
        print("i am hod")
h= HOD()
print(h.hod_info())
h.teacher_info()
h.person_info()
```

```python
class Developer:
    def developer_info(a):
        print("its developr")
class Manager:
    def manager_info(b):
        print("its manger")
class Teamleader(Developer,Manager):
    def teamleader_show(a):
        print("its teamleader")
a = Teamleader()
print(a.developer_info())
print(a.manager_info())
print(a.teamleader_show())
```

```python
#heriarchical inheritance

class Developer:
    def developer(d):
        print("coding")
class Manager(Developer):
    def manager(m):
        print("managing")
class Teamleader(Developer):
    def teamleader(t):
        print("leading")
a = Manager()
b=Teamleader()
c = Developer()
print("a.coding")
print("b.leading")
```

```python
#hybrid inheritance


class Person:
    def person(p):
        print("muskan")
class Teacher(Person):
    def teacher(t):
        print("teaches")
class Student(Person):
    def student(s):
        print("study")
class HOD(Teacher,Student):
    def hod(h):
        print("heirs")
a = HOD()
b = Person()
a.student()
a.teacher()
a.hod()
b.person()
a.person()
```
