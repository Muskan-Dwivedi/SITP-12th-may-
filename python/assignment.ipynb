8.Create a class FactorialCalculator with a function to calculate factorial of a number using a loop


```python
class FactorialCalculator:
    def calculate_factorial(self, number):
        factorial = 1

        # Check for negative numbers
        if number < 0:
            return "Factorial does not exist for negative numbers"

        # Calculate factorial using loop
        for i in range(1, number + 1):
            factorial *= i

        return factorial


# Create object
fc = FactorialCalculator()

# Example usage
num = 5
print(f"Factorial of {num} is:", fc.calculate_factorial(num))
```
