# class in python

In Python, a class is a blueprint or template for creating objects. It bundles data (attributes) and behavior (methods) into a single unit, which is the core of Object-Oriented Programming (OOP).

#### four core pillars: 
Encapsulation

 Abstraction

 Inheritance

 Polymorphism.
 
  These principles help organize code into modular, reusable, and maintainable structures.
         
  ### Python Class Notes
| Topic                 | Description                                                        |
| --------------------- | ------------------------------------------------------------------ |
| **Class**             | A class is a blueprint used to create objects.                     |
| **Object**            | An object is an instance of a class.                               |
| **OOP**               | OOP stands for Object-Oriented Programming.                        |
| **Syntax**            | `class ClassName:`                                                 |
| **Constructor**       | `__init__()` is a special method used to initialize object values. |
| **self Keyword**      | `self` refers to the current object of the class.                  |
| **Method**            | A function inside a class is called a method.                      |
| **Attribute**         | Variables inside a class are called attributes.                    |
| **Instance Variable** | Variables created using `self` are instance variables.             |
| **Object Creation**   | `obj = ClassName()`                                                |
| **Advantages**        | Code reusability, better structure, easy maintenance.              |



#### Four pillar of OOPS

| **Pillar**             | **Definition**                |  **Example**         |
|----------------------|----------------------------    |-----------------------|
| Encapsulation        | weapping data and method into sinlge unit           | class and object  |
| Inheritance          | one class acquire properties of another class       | parent and child  |
| Polymorphism         | same function behave diffrently at different situation  | Method overriding |
| Data Abstraction     | hidng internal detail and showing essential infomation  | Astract class    |
