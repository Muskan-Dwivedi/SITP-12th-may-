2. Create a class Calculator with methods for addition, subtraction, multiplication, and division.

```python
class Calculator:
    def __init__(self):
     self.a = int(input("enter the first number:"))
     self.b = int(input("enter the second number:"))
    def addition(self):
        self.c = self.a + self.b
        print("addition:", self.c)
    def subtraction(self):
        self.d = self.a - self.b
        print("subtraction:", self.d)
    def multiplication(self):
       self.e = self.a * self.b
       print("multiplication:", self.e)
    def division(self):
       self.f = self.a / self. b
       print("division :", self.f)

cal = Calculator()
cal.addition()
cal.subtraction()
cal.multiplication()
cal.division()
```

4. Create a class Vehicle with method display(). Inherit it into class Car and call parent method.

```python
class Vehical:
    vehical_name = "Car"
    def display(self):
        print("name of vehical is :",self.vehical_name)

class Car(Vehical):
    car_name = " Mahindra Scorpio"
    car_modal = "S11"
    car_price = 1700000
    
    def show(self):
        print("name of car is :", self.car_name)
        print("modal of car is :", self.car_modal)
        print("price of car is :", self.car_price)
c = Car()
c.show()
```
