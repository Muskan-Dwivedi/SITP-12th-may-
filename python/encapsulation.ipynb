## Encapsulation

Encapsulation in Python is the process of bundling data (attributes) and the methods (functions) that operate on that data into a single unit, usually a class.



```python
class Collage:
    collage_name= "xyz"       #public variable
    collage_garden="yes"      #public variable
    collage_account="27437324"    #public variable

clg = Collage()
clg.collage_garden
```

```python
clg.collage_account
```

```python
class Collage:
    collage_name= "xyz"       #public variable
    collage_garden="yes"      #public variable
    __collage_account="27437324"    #private variable(double underscore means private)

clg = Collage()
```

```python
clg.__collage_account
```

```python
class Collage:
    collage_name= "xyz"       #public variable
    _collage_garden="yes"      #proteced variable(sinlge underscore means proteced)
    collage_account="27437324"    #public variable

clg = Collage()
clg._collage_garden
```

```python
clg._collage_garden
```

```python
class Collage:
    collage_name= "xyz"       #public variable
    collage_garden="yes"      #public variable
    __collage_account="27437324"    #private variable(double underscore means private)
    def display(self):
        return self.__collage_account

clg = Collage()
```

```python
clg.display()
```

```python
class Collage:
    collage_name= "xyz"       
    collage_garden="yes"      
    __collage_account="27437324"    #private variable(double underscore means private)
    def display(self,a,b):              ####self can be keyword or object, and parameter  
        print(a+b)
        return self.__collage_account

clg = Collage()
```

```python
clg.display( a = 12,b = 12)
```
