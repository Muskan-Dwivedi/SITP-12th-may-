### Constructor

A constructor is a special method in Python that is automatically called when an object of a class is created.
It is mainly used to initialize the object's data members (attributes).

In Python, the constructor method is written as __init__().

class ClassName:
    def __init__(self):
        # initialization code

### Types of Constructors in Python


|Type of Constructor  |	Description	  |   Example   |  
|---------------------|---------------|-------------|  
|  default Constructor |	A constructor with no parameters except self. It initializes objects with default values.|	    python class Student: def __init__(self): print("Default Constructor Called") s1 = Student()|
| Parameterized Constructor  |	A constructor that accepts parameters to initialize object values.|	python class Student: def __init__(self, name, age): self.name = name self.age = age s1 = Student("Muskan", 20)|
        

Key Points
|Point	         |Explanation|
|----------------|------------------------|
|Special Method	 | Constructor uses __init__() method|
|Automatic Call|	Called automatically when object is created|
|Purpose |	Initializes object attributes|
|First Parameter	 |self refers to the current object|
Multiple Constructors |	Python does not support multiple constructors directly|
