```python
try:
    num = int(input("enter the number"))
except ValueError as e:
    print("invalid input",e)
else:
    print("your input",num)
```

```python
class InvalidBalance(Exception):
    pass 

try:
    bal = 50000
    withdrawal =float(input("enter the amount"))
    if withdrawal>bal:
        raise InvalidBalance("withdrawal not possile")
    print("possible withdrawal")
except InvalidBalance as e:
    print(e)

```

```python
class InvalidPassLength(Exception):
    pass 

try:
    length = 5
    password = input("enter the password")
    if len(password)!= length:
        raise InvalidPassLength("length of password is not matched")
    if password != password.istitle():
        raise InvalidPassLength("First letter should be capital")
    print("password is:",password)        
except InvalidPassLength as e:
    print(e)
    
```
