File handling in Python means working with files to create, read, write, append, and delete data stored on a computer.

A file is used to store data permanently, unlike variables that store data temporarily in memory.

Types of File Operations
Operation	Description
Open	Opens a file
Read	Reads data from a file
Write	Writes data into a file
Append	Adds new data at the end
Close	Closes the file
Syntax of File Handling
file = open("filename", "mode")
"filename" → Name of the file
"mode" → Purpose of opening the file



| Mode   | Meaning           |
| ------ | ----------------- |
| `"r"`  | Read mode         |
| `"w"`  | Write mode        |
| `"a"`  | Append mode       |
| `"x"`  | Create new file   |
| `"rb"` | Read binary file  |
| `"wb"` | Write binary file |
